## Pemrosesan Dataset Menggunakan PySpark

Pada tahap ini dataset dibaca menggunakan RDD PySpark.
Data kemudian akan dipersiapkan untuk proses preprocessing dan WordCount.

```python
# Mengaktifkan kembali SparkSession

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("WordCount_BigData") \
    .getOrCreate()

print("SparkSession berhasil diaktifkan.")
print("Versi Spark :", spark.version)
```

```python
# Membaca dataset menggunakan PySpark RDD

sumber_data = "bigdata.txt"

rdd_teks = spark.sparkContext.textFile(sumber_data)

print("Dataset berhasil dibaca menggunakan PySpark.")
```

```python
# Menghitung jumlah baris dataset melalui RDD

total_data_spark = rdd_teks.count()

print("Jumlah baris yang dibaca PySpark :", total_data_spark)
```

```python
# Mengambil 5 contoh baris dari RDD

contoh_data_spark = rdd_teks.take(5)

print("Contoh data yang dibaca PySpark:\n")

for nomor, baris in enumerate(contoh_data_spark, start=1):
    print(f"{nomor}. {baris}")
```

```python
# Melihat jumlah partition awal pada RDD

partition_awal = rdd_teks.getNumPartitions()

print("Jumlah partition awal :", partition_awal)
```

```python
# Fungsi preprocessing untuk membersihkan setiap baris teks

import re

def bersihkan_teks(teks):
    teks = teks.lower()
    teks = re.sub(r"[^a-z0-9\s]", " ", teks)
    teks = re.sub(r"\s+", " ", teks).strip()
    return teks
```

```python
# Menerapkan preprocessing pada seluruh dataset

rdd_bersih = rdd_teks.map(bersihkan_teks)

print("Proses preprocessing berhasil didefinisikan.")
```

```python
# Menampilkan 5 contoh data setelah preprocessing

contoh_data_bersih = rdd_bersih.take(5)

print("Contoh data setelah preprocessing:\n")

for nomor, teks in enumerate(contoh_data_bersih, start=1):
    print(f"{nomor}. {teks}")
```

```python
# Memastikan jumlah data tetap sama setelah preprocessing

jumlah_setelah_preprocessing = rdd_bersih.count()

print("Jumlah data sebelum preprocessing :", total_data_spark)
print("Jumlah data setelah preprocessing :", jumlah_setelah_preprocessing)
```

```python
# Memecah setiap baris teks menjadi kata-kata individual

rdd_kumpulan_kata = rdd_bersih.flatMap(
    lambda baris_teks: baris_teks.split()
)

print("Proses tokenisasi dengan flatMap() berhasil didefinisikan.")
```

```python
# Menampilkan 20 kata pertama hasil tokenisasi

contoh_kata = rdd_kumpulan_kata.take(20)

print("20 kata pertama setelah tokenisasi:\n")

for nomor, kata in enumerate(contoh_kata, start=1):
    print(f"{nomor}. {kata}")
```

```python
# Menghitung total kata pada seluruh dataset

total_kata_dataset = rdd_kumpulan_kata.count()

print("Jumlah baris dataset :", jumlah_setelah_preprocessing)
print("Total kata dataset   :", total_kata_dataset)
```

```python
# Memeriksa apakah terdapat token kosong setelah proses pemisahan kata

jumlah_kata_kosong = rdd_kumpulan_kata.filter(
    lambda kata: kata.strip() == ""
).count()

print("Jumlah token kosong :", jumlah_kata_kosong)

if jumlah_kata_kosong == 0:
    print("Status              : Tokenisasi bersih dan siap untuk WordCount.")
else:
    print("Status              : Masih terdapat token kosong.")
```

```python
# Mengubah setiap kata menjadi pasangan (kata, 1)

rdd_pasangan_kata = rdd_kumpulan_kata.map(
    lambda kata: (kata, 1)
)

print("Pembentukan pasangan (kata, 1) berhasil didefinisikan.")
```

```python
# Menampilkan 15 pasangan kata pertama

contoh_pasangan = rdd_pasangan_kata.take(15)

print("Contoh pasangan (kata, 1):\n")

for nomor, pasangan in enumerate(contoh_pasangan, start=1):
    print(f"{nomor}. {pasangan}")
```

```python
# Memastikan setiap kata telah berubah menjadi satu pasangan (kata, 1)

total_pasangan_kata = rdd_pasangan_kata.count()

print("Total kata sebelum map() :", total_kata_dataset)
print("Total pasangan setelah map() :", total_pasangan_kata)

if total_pasangan_kata == total_kata_dataset:
    print("Status : Setiap kata berhasil diubah menjadi pasangan (kata, 1).")
else:
    print("Status : Terdapat perbedaan jumlah data.")
```

```python
# Menghitung frekuensi setiap kata menggunakan reduceByKey()

rdd_frekuensi_kata = rdd_pasangan_kata.reduceByKey(
    lambda nilai_a, nilai_b: nilai_a + nilai_b
)

print("Proses WordCount dengan reduceByKey() berhasil didefinisikan.")
```

```python
# Menampilkan 15 hasil WordCount pertama

contoh_hasil_wordcount = rdd_frekuensi_kata.take(15)

print("Contoh hasil WordCount:\n")

for nomor, hasil in enumerate(contoh_hasil_wordcount, start=1):
    print(f"{nomor}. {hasil[0]} = {hasil[1]}")
```

```python
# Menghitung jumlah kata unik

jumlah_kata_unik = rdd_frekuensi_kata.count()

print("Total kata dalam dataset :", total_kata_dataset)
print("Jumlah kata unik         :", jumlah_kata_unik)
```

```python
# Memastikan total hasil WordCount sama dengan total kata awal

total_frekuensi = rdd_frekuensi_kata.map(
    lambda pasangan: pasangan[1]
).sum()

print("Total kata hasil tokenisasi :", total_kata_dataset)
print("Total frekuensi WordCount   :", total_frekuensi)

if total_frekuensi == total_kata_dataset:
    print("Status : Hasil WordCount berhasil diverifikasi.")
else:
    print("Status : Terdapat ketidaksesuaian hasil.")
```

```python
# Mengurutkan hasil WordCount berdasarkan frekuensi tertinggi

rdd_kata_terurut = rdd_frekuensi_kata.sortBy(
    lambda data: (-data[1], data[0])
)

print("Pengurutan frekuensi kata berhasil didefinisikan.")
```

```python
# Mengambil 10 kata dengan frekuensi kemunculan tertinggi

sepuluh_kata_teratas = rdd_kata_terurut.take(10)

print("10 KATA DENGAN FREKUENSI TERTINGGI")
print("-" * 42)

for peringkat, (kata, frekuensi) in enumerate(sepuluh_kata_teratas, start=1):
    print(f"{peringkat:2}. {kata:<15} = {frekuensi}")
```

```python
# Menyiapkan fungsi untuk menguji waktu WordCount
# pada jumlah partition yang berbeda

import time

def uji_partition(jumlah_partition):

    # Membagi ulang dataset sesuai jumlah partition yang diuji
    rdd_uji = rdd_bersih.repartition(jumlah_partition)

    # Menyimpan sementara agar proses repartition
    # tidak tercampur dengan pengukuran WordCount
    rdd_uji.cache()
    rdd_uji.count()

    waktu_mulai = time.perf_counter()

    hasil_uji = (
        rdd_uji
        .flatMap(lambda baris: baris.split())
        .map(lambda kata: (kata, 1))
        .reduceByKey(lambda a, b: a + b)
    )

    # Action untuk memicu proses Spark
    jumlah_kata_uji = hasil_uji.count()

    waktu_selesai = time.perf_counter()

    durasi = waktu_selesai - waktu_mulai

    # Membersihkan cache setelah pengujian
    rdd_uji.unpersist()

    return jumlah_kata_uji, durasi
```

```python
# Pengujian WordCount menggunakan 1 partition

kata_unik_1, waktu_1 = uji_partition(1)

print("PENGUJIAN 1 PARTITION")
print("----------------------------")
print("Jumlah partition :", 1)
print("Kata unik        :", kata_unik_1)
print("Waktu proses     :", round(waktu_1, 4), "detik")
```

```python
# Pengujian WordCount menggunakan 2 partition

kata_unik_2, waktu_2 = uji_partition(2)

print("PENGUJIAN 2 PARTITION")
print("----------------------------")
print("Jumlah partition :", 2)
print("Kata unik        :", kata_unik_2)
print("Waktu proses     :", round(waktu_2, 4), "detik")
```

```python
# Pengujian WordCount menggunakan 4 partition

kata_unik_4, waktu_4 = uji_partition(4)

print("PENGUJIAN 4 PARTITION")
print("----------------------------")
print("Jumlah partition :", 4)
print("Kata unik        :", kata_unik_4)
print("Waktu proses     :", round(waktu_4, 4), "detik")
```

```python
# Pengujian WordCount menggunakan 8 partition

kata_unik_8, waktu_8 = uji_partition(8)

print("PENGUJIAN 8 PARTITION")
print("----------------------------")
print("Jumlah partition :", 8)
print("Kata unik        :", kata_unik_8)
print("Waktu proses     :", round(waktu_8, 4), "detik")
```

```python
# Membuat tabel perbandingan waktu proses

import pandas as pd

hasil_partition = pd.DataFrame({
    "Jumlah Partition": [1, 2, 4, 8],
    "Jumlah Kata Unik": [
        kata_unik_1,
        kata_unik_2,
        kata_unik_4,
        kata_unik_8
    ],
    "Waktu Proses (detik)": [
        waktu_1,
        waktu_2,
        waktu_4,
        waktu_8
    ]
})

hasil_partition["Waktu Proses (detik)"] = (
    hasil_partition["Waktu Proses (detik)"].round(4)
)

display(hasil_partition)
```

```python
# Menentukan konfigurasi partition dengan waktu tercepat

baris_tercepat = hasil_partition.loc[
    hasil_partition["Waktu Proses (detik)"].idxmin()
]

print("HASIL PENGUJIAN TERCEPAT")
print("----------------------------")
print(
    "Jumlah partition :",
    int(baris_tercepat["Jumlah Partition"])
)
print(
    "Waktu proses     :",
    baris_tercepat["Waktu Proses (detik)"],
    "detik"
)
```

```python
# Membuat pipeline WordCount khusus untuk analisis Job, Stage, dan Task

rdd_demo_6a = (
    rdd_bersih
    .flatMap(lambda baris: baris.split())
    .map(lambda kata: (kata, 1))
    .reduceByKey(lambda nilai1, nilai2: nilai1 + nilai2)
)

print("Pipeline analisis berhasil dibuat.")
print("Partition sebelum reduceByKey :", rdd_bersih.getNumPartitions())
print("Partition hasil reduceByKey   :", rdd_demo_6a.getNumPartitions())
```

```python
# Melihat lineage atau alur ketergantungan RDD

lineage_rdd = rdd_demo_6a.toDebugString()

if isinstance(lineage_rdd, bytes):
    lineage_rdd = lineage_rdd.decode("utf-8")

print("LINEAGE RDD WORDCOUNT")
print("------------------------------")
print(lineage_rdd)
```

```python
# Menjalankan Action untuk memicu Spark Job

sc = spark.sparkContext

nama_job = "analisis_wordcount_6a"

sc.setJobGroup(
    nama_job,
    "Analisis Job WordCount Tahap 6A"
)

hasil_wordcount_6a = rdd_demo_6a.count()

job_ids = sc.statusTracker().getJobIdsForGroup(nama_job)

print("HASIL EKSEKUSI JOB")
print("------------------------------")
print("Jumlah kata unik :", hasil_wordcount_6a)
print("Job ID           :", job_ids)
```

```python
# Menampilkan informasi Stage dan Task dari Job

# Memastikan sc (SparkContext) terdefinisi dari SparkSession aktif
sc = spark.sparkContext
tracker = sc.statusTracker()

# Memastikan job_ids terdefinisi dari group aktif jika sebelumnya hilang
try:
    job_ids
except NameError:
    job_ids = sc.statusTracker().getJobIdsForGroup("analisis_wordcount_6a")

print("DETAIL JOB, STAGE, DAN TASK")
print("----------------------------------------")

for job_id in job_ids:

    info_job = tracker.getJobInfo(job_id)

    if info_job is not None:

        stage_ids = list(info_job.stageIds)

        print(f"\nJob ID    : {job_id}")
        print(f"Stage ID  : {stage_ids}")
        print(f"Jumlah Stage : {len(stage_ids)}")

        for stage_id in stage_ids:

            info_stage = tracker.getStageInfo(stage_id)

            if info_stage is not None:

                print(f"\n  Stage {stage_id}")
                print(f"  Jumlah Task       : {info_stage.numTasks}")
                print(f"  Task Selesai      : {info_stage.numCompletedTasks}")
                print(f"  Task Gagal        : {info_stage.numFailedTasks}")
```
